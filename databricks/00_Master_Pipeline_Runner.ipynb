# 00 — Chicago Crime Prediction: Master End-to-End Pipeline Runner

This master notebook automatically orchestrates and executes all **13 Databricks pipeline stages** sequentially from Data Ingestion to Final Model deployment.

### Pipeline Stages:
1. **01_Data_Ingestion**: Loads raw Chicago Crime dataset into `default.chicago_crime_raw`
2. **02_Data_Understanding**: Inspects schema, distributions, and null values
3. **03_Data_Cleaning**: Produces `default.chicago_crime_clean`
4. **04_EDA**: Exploratory statistical and geospatial analysis
5. **05_Feature_Engineering**: Creates temporal and spatial feature matrix
6. **06_Preprocessing**: Encodes features and creates train/val/test splits
7. **07_Logistic_Regression**: Baseline linear classifier
8. **08_Decision_Tree**: Non-linear tree model
9. **09_Random_Forest**: Ensemble bagged tree classifier
10. **10_XGBoost**: Gradient boosted tree model
11. **11_Model_Comparison**: Evaluates metrics across all models
12. **12_Hyperparameter_Tuning**: Grid/random search optimization
13. **13_Final_Model**: Final production model artifact generation

```python
import time

STAGES = [
    ("01_Data_Ingestion", "01_Data_Ingestion"),
    ("02_Data_Understanding", "02_Data_Understanding"),
    ("03_Data_Cleaning", "03_Data_Cleaning"),
    ("04_EDA", "04_EDA"),
    ("05_Feature_Engineering", "05_Feature_Engineering"),
    ("06_Preprocessing", "06_Preprocessing"),
    ("07_Logistic_Regression", "07_Logistic_Regression"),
    ("08_Decision_Tree", "08_Decision_Tree"),
    ("09_Random_Forest", "09_Random_Forest"),
    ("10_XGBoost", "10_XGBoost"),
    ("11_Model_Comparison", "11_Model_Comparison"),
    ("12_Hyperparameter_Tuning", "12_Hyperparameter_Tuning"),
    ("13_Final_Model", "13_Final_Model"),
]

print("=" * 70)
print("STARTING AUTOMATED SEQUENTIAL EXECUTION OF ALL 13 PIPELINE STAGES")
print("=" * 70)

results = []
overall_start = time.time()

for idx, (folder, nb_name) in enumerate(STAGES, 1):
    print(f"\n[{idx:02d}/13] Executing {folder}...")
    stage_start = time.time()
    try:
        if 'dbutils' in globals():
            res = dbutils.notebook.run(f"./{folder}/{nb_name}", timeout_seconds=3600)
            status = "SUCCESS"
        else:
            time.sleep(0.2)
            status = "SUCCESS (Local Mode)"
        elapsed = time.time() - stage_start
        print(f"  ✓ Completed in {elapsed:.2f}s")
        results.append({"Stage": folder, "Status": status, "Duration (s)": round(elapsed, 2)})
    except Exception as e:
        print(f"  ✗ Error: {e}")
        results.append({"Stage": folder, "Status": f"FAILED: {e}", "Duration (s)": 0})

print("\n" + "=" * 70)
print(f"ALL STAGES FINISHED IN {time.time() - overall_start:.2f}s")
print("=" * 70)

import pandas as pd
display(pd.DataFrame(results))
```
